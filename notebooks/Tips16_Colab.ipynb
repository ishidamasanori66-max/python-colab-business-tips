**業務の困りごと**：商品ページのURLごとにQR画像を作る。

**このTipsでできること**：`qrcode.make` で各URLのQR画像を生成し、商品名をファイル名にします。

`/content/` はColabの一時領域です。PCやGoogle Driveには自動保存されません。左側のフォルダから結果を選んでダウンロードできます。セッション終了前に保存してください。

上から順番に実行してください。各Tipsのサンプルはノートブック内で作ります。


## 1. 練習用データと準備

商品名と練習用URLを2組用意します。`example.com` は実際の配布先ではありません。

最後の `display()` で練習用データ全体を表として表示します。列名・行数・値を確認してから次へ進みましょう。


```python
try:
    from IPython.display import display
except ImportError:
    display = print  # Colab以外の環境向け
%pip -q install 'qrcode[pil]'
import qrcode
from pathlib import Path
base=Path('/content') if Path('/content').exists() else Path.cwd()
# 商品名とQRに記録するURLの対応表
items={'商品A':'https://example.com/a','商品B':'https://example.com/b'}
import pandas as pd
print('QRコード化するデータ（先頭5行）:')
display(pd.DataFrame(items.items(),columns=['商品','URL']))
```

## 2. 自動処理

`qrcode.make` で各URLのQR画像を生成し、商品名をファイル名にします。

**結果を見るポイント**：画像2枚が作られ、ZIPでまとめて保存できます。配布前にスマートフォンで読み取り先を確認してください。

保存した成果物を読み直して `display()` で表示します。画面の表とダウンロードしたファイルの内容を照らし合わせてください。


```python
try:
    from IPython.display import display
except ImportError:
    display = print  # Colab以外の環境向け
# 出力先フォルダを作る
folder=base/'tips16_QR'; folder.mkdir(exist_ok=True)
# URLごとにQR画像を生成する
for label,url in items.items():
    qrcode.make(url).save(folder/f'{label}.png')
print(list(folder.glob('*.png')))
# 複数の成果物をPCでまとめて扱えるようZIPにする
import shutil
out = base / 'Tips16_成果物.zip'
shutil.make_archive(str(out.with_suffix('')), 'zip', root_dir=folder)
print('ダウンロード用ZIP:', out)

# 作成したQR画像のファイル名と対応URLを表示する
display(pd.DataFrame([{'ファイル名': f'{label}.png', 'URL': url}
                      for label, url in items.items()]))
```

## 3. 結果の確認

`assert` は条件が正しいか確認する命令です。条件が違えばそこで止まるため、想定外の集計を見つけられます。

画像2枚が作られ、ZIPでまとめて保存できます。配布前にスマートフォンで読み取り先を確認してください。


```python
# 件数や合計が期待値と一致するか検査する
assert len(list(folder.glob('*.png')))==len(items)
print('確認OK')
assert out.exists()  # ダウンロード用ZIPも確認する
```

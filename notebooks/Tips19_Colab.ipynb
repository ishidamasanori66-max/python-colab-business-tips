# Tips 19：Google Driveのファイルを処理

**業務の困りごと**：Drive上のCSVを開いて集計する。

**このTipsでできること**：店舗別に売上を合計し、CSVに保存します。発展セルでは本人が認証したDriveの練習用フォルダに結果をコピーします。

**使用ライブラリ**：google.colab.drive, pandas

### ファイルの場所
`/content/` はColabの一時領域です。PCやGoogle Driveには自動保存されません。左側のフォルダから結果を選んでダウンロードできます。セッション終了前に保存してください。

上から順番に実行してください。各Tipsのサンプルはノートブック内で作ります。


## 1. 練習用データと準備

店舗別売上のCSVをColabの一時領域に作ります。この段階ではDriveに保存されていません。

最後の `head()` で先頭の行を表として表示します。列名・行数・値を確認してから次へ進みましょう。


```python
import pandas as pd
from pathlib import Path
base=Path('/content') if Path('/content').exists() else Path.cwd()
# Colabの一時領域にCSVを作る
src=base/'tips19_売上.csv'
pd.DataFrame({'店舗':['鹿児島','霧島'],'売上':[1200,900]}).to_csv(src,index=False,encoding='utf-8-sig')
print('まずは一時領域:',src)
print('練習用CSV（先頭5行）:')
pd.read_csv(src,encoding='utf-8-sig').head()
```

## 2. 自動処理

店舗別に売上を合計し、CSVに保存します。発展セルでは本人が認証したDriveの練習用フォルダに結果をコピーします。

**結果を見るポイント**：2店舗の合計は **2,100円** です。発展セルは認証した場合にだけ実行してください。


```python
# CSVを読み込む
frame=pd.read_csv(src,encoding='utf-8-sig')
# 店舗別の集計結果を別のCSVへ出力する
out=base/'Tips19_売上集計.csv'
summary = frame.groupby('店舗', as_index=False)['売上'].sum()
summary.to_csv(out, index=False, encoding='utf-8-sig')
print(pd.read_csv(out))
```

## 3. 結果の確認

`assert` は条件が正しいか確認する命令です。条件が違えばそこで止まるため、想定外の集計を見つけられます。

2店舗の合計は **2,100円** です。発展セルは認証した場合にだけ実行してください。


```python
# 件数や合計が期待値と一致するか検査する
assert out.exists() and pd.read_csv(out)['売上'].sum()==2100
print('確認OK')
```

## 発展：Google Driveへ保存（Colab専用・任意）
認証画面で本人のGoogleアカウントを選びます。`MyDrive/ColabTips` を作成してコピーします。


```python
# この発展セルはColabで本人のGoogle認証を行う場合だけ実行
# 先に通常の処理セルを実行し、out に集計結果を保存しておく
from google.colab import drive
drive.mount('/content/drive')
import shutil
destination=Path('/content/drive/MyDrive/ColabTips')
destination.mkdir(exist_ok=True)
shutil.copy2(out,destination/out.name)
print('Driveへ保存:',destination/out.name)
```

## 4. 持ち帰り・練習

**練習問題**：Drive接続セルを実行し、MyDriveの自分専用フォルダへ結果をコピーしてみましょう。

次のセルはColab上で本人の認証後にだけ実行してください。

**実務で使う前に**：Driveに保存するとセッション終了後もファイルが残ります。

出力をPCへ保存したい場合は次のセルをColabで実行します。Colab以外では出力場所を表示します。

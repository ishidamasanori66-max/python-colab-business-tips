# Tips 07：売上グラフを自動作成

**業務の困りごと**：月次売上のグラフを毎回手で描く。

**このTipsでできること**：`plot` で点を結んだ折れ線を描き、`savefig` でPNGを保存します。横軸が月、縦軸が売上です。

**使用ライブラリ**：pandas, matplotlib

### ファイルの場所
`/content/` はColabの一時領域です。PCやGoogle Driveには自動保存されません。左側のフォルダから結果を選んでダウンロードできます。セッション終了前に保存してください。

上から順番に実行してください。各Tipsのサンプルはノートブック内で作ります。


## 1. 練習用データと準備

7〜9月の月別売上を用意します。月の順番と金額の対応を確認してください。

最後の `head()` で先頭の行を表として表示します。列名・行数・値を確認してから次へ進みましょう。


```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
base=Path('/content') if Path('/content').exists() else Path.cwd()
# 月と売上の対応を表にする
data=pd.DataFrame({'月':['7月','8月','9月'],'売上':[220000,260000,240000]})
print('練習用データ（先頭5行）:')
data.head()
```

## 2. 自動処理

`plot` で点を結んだ折れ線を描き、`savefig` でPNGを保存します。横軸が月、縦軸が売上です。

**結果を見るポイント**：8月が最大の **260,000円** になっているか、画像でも確かめます。


```python
# グラフのキャンバスと描画領域を作る
fig,ax=plt.subplots(figsize=(7,4))
# 月別売上を点付きの折れ線で描く
ax.plot(data['月'],data['売上'],marker='o')
ax.set(title='Monthly sales',xlabel='Month',ylabel='Sales (JPY)')
ax.grid(alpha=.3)
# 画像ファイルとして保存する
out=base/'Tips07_売上推移.png'; fig.tight_layout(); fig.savefig(out,dpi=150)
plt.show(); print(out)
```

## 3. 結果の確認

`assert` は条件が正しいか確認する命令です。条件が違えばそこで止まるため、想定外の集計を見つけられます。

8月が最大の **260,000円** になっているか、画像でも確かめます。


```python
# 件数や合計が期待値と一致するか検査する
assert out.is_file() and out.stat().st_size>0
print('確認OK')
```

# Tips 03：CSVをExcelへ変換

**業務の困りごと**：CSVを開いて手作業でExcel形式に保存する。

**このTipsでできること**：`read_csv` でCSVを表に読み込み、`to_excel` で `.xlsx` を作ります。`index=False` はPythonの行番号を余計な列として出さない指定です。

**使用ライブラリ**：pandas, openpyxl

### ファイルの場所
`/content/` はColabの一時領域です。PCやGoogle Driveには自動保存されません。左側のフォルダから結果を選んでダウンロードできます。セッション終了前に保存してください。

上から順番に実行してください。各Tipsのサンプルはノートブック内で作ります。


## 1. 練習用データと準備

2行のCSVを作ります。`utf-8-sig` はWindowsのExcelなどで日本語を開きやすくするための指定です。

最後の `display()` で練習用データ全体を表として表示します。列名・行数・値を確認してから次へ進みましょう。


```python
try:
    from IPython.display import display
except ImportError:
    display = print  # Colab以外の環境向け
import pandas as pd
from pathlib import Path
base=Path('/content') if Path('/content').exists() else Path.cwd()
# 元になるCSVの場所
src=base/'tips03_売上.csv'
# 日本語を含む練習用CSVを作る
pd.DataFrame({'商品':['茶','菓子'],'売上金額':[1200,850]}).to_csv(src,index=False,encoding='utf-8-sig')
print(src)
print('練習用CSV（先頭5行）:')
display(pd.read_csv(src,encoding='utf-8-sig'))
```

## 2. 自動処理

`read_csv` でCSVを表に読み込み、`to_excel` で `.xlsx` を作ります。`index=False` はPythonの行番号を余計な列として出さない指定です。

**結果を見るポイント**：変換前後の「商品」「売上金額」の値が同じであることを確認します。

保存した成果物を読み直して `display()` で表示します。画面の表とダウンロードしたファイルの内容を照らし合わせてください。


```python
try:
    from IPython.display import display
except ImportError:
    display = print  # Colab以外の環境向け
# CSVを表に読み込む
df=pd.read_csv(src,encoding='utf-8-sig')
# 出力先は拡張子 .xlsx にする
out=base/'Tips03_売上.xlsx'
# 余計な行番号を付けずに保存する
df.to_excel(out,index=False)
print(df)

# 保存後のExcelを読み直して表示する
display(pd.read_excel(out))
```

## 3. 結果の確認

`assert` は条件が正しいか確認する命令です。条件が違えばそこで止まるため、想定外の集計を見つけられます。

変換前後の「商品」「売上金額」の値が同じであることを確認します。


```python
# 件数や合計が期待値と一致するか検査する
assert pd.read_excel(out).equals(df)
print('確認OK:',out)
```

## 4. 持ち帰り・練習

**練習問題**：CSVの列を追加して、Excelに反映されるか試しましょう。

実データがShift-JIS系なら encoding="cp932" を試します。文字化けを確認してください。

**実務で使う前に**：実務のCSVが文字化けする場合、元ファイルの文字コードを調べてから `encoding` を変更します。

出力をPCへ保存したい場合は次のセルをColabで実行します。Colab以外では出力場所を表示します。


```python
# Colab上で作った成果物をPCへダウンロードする
try:
    from google.colab import files
    files.download(str(out))
except ImportError:
    print('出力ファイル:',out.resolve())
```

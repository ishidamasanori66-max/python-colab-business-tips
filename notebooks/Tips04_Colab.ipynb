**このTipsでできること**：`dropna` で必須項目が空の行を除き、`drop_duplicates` で顧客IDの重複を除きます。最後に `query` で有効な顧客だけを残します。

## 1. 練習用データと準備

欠損した行、同じ顧客IDが重複した行、退会済みの行を含むCSVです。

最後の `display()` で練習用データ全体を表として表示します。列名・行数・値を確認してから次へ進みましょう。


```python
try:
    from IPython.display import display
except ImportError:
    display = print  # Colab以外の環境向け
import pandas as pd
from pathlib import Path
base=Path('/content') if Path('/content').exists() else Path.cwd()
# 元データのCSVを保存する
src=base/'tips04_顧客.csv'
# 欠損・重複・退会済みの行を混ぜる
pd.DataFrame({'顧客ID':[1,2,2,3,None],'氏名':['田中','佐藤','佐藤','山田',None],'状態':['有効','有効','有効','退会',None]}).to_csv(src,index=False,encoding='utf-8-sig')
print('重複や欠損を含む元データ（練習用データ）:')
display(pd.read_csv(src,encoding='utf-8-sig'))
```

## 2. 自動処理

`dropna` で必須項目が空の行を除き、`drop_duplicates` で顧客IDの重複を除きます。最後に `query` で有効な顧客だけを残します。

**結果を見るポイント**：入力は5行、出力は **2行** です。重複の判定キーは業務で一意になる列を選びましょう。

保存した成果物を読み直して `display()` で表示します。画面の表とダウンロードしたファイルの内容を照らし合わせてください。


```python
try:
    from IPython.display import display
except ImportError:
    display = print  # Colab以外の環境向け
# 元のCSVを読み込む
df=pd.read_csv(src,encoding='utf-8-sig')
# 必須項目の欠損とIDの重複を取り除き、有効行を選ぶ
clean = (
    df.dropna(subset=['顧客ID', '氏名'])
      .drop_duplicates(subset=['顧客ID'])
      .query('状態 == "有効"')
      .copy()
)
# 欠損を除いたIDを整数に戻す
clean['顧客ID']=clean['顧客ID'].astype(int)
out = base / 'Tips04_有効顧客.csv'
clean.to_csv(out, index=False, encoding='utf-8-sig')
print('処理前:', len(df), '件 / 処理後:', len(clean), '件')
print(clean)

# 保存したCSVを読み直して表示する
display(pd.read_csv(out, encoding='utf-8-sig'))
```

## 3. 結果の確認

`assert` は条件が正しいか確認する命令です。条件が違えばそこで止まるため、想定外の集計を見つけられます。

入力は5行、出力は **2行** です。重複の判定キーは業務で一意になる列を選びましょう。


```python
# 件数や合計が期待値と一致するか検査する
assert len(clean)==2 and clean['顧客ID'].is_unique
print('確認OK:',out)
```

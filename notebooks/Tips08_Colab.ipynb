## 1. 練習用データと準備

10件の満足度（1〜5点）と自由記述を作ります。

最後の `display()` で練習用データ全体を表として表示します。列名・行数・値を確認してから次へ進みましょう。


```python
try:
    from IPython.display import display
except ImportError:
    display = print  # Colab以外の環境向け
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
base=Path('/content') if Path('/content').exists() else Path.cwd()
# 満足度と自由記述の練習用回答を作る
df=pd.DataFrame({'満足度':[5,4,3,5,4,2,5,3,4,5], '自由記述':['良い','便利','普通','助かる','良い','難しい','良い','普通','便利','良い']})
# 回答をCSVへ保存する
src=base/'tips08_アンケート.csv'; df.to_csv(src,index=False,encoding='utf-8-sig')
print('アンケートの元データ（練習用データ）:')
display(pd.read_csv(src,encoding='utf-8-sig'))
```

## 2. 自動処理

`value_counts` は各点数の件数を数えます。「割合」は各件数を回答総数で割り、棒グラフとExcelに書き出します。

**結果を見るポイント**：件数の合計は **10件**、割合の合計は **1.0（100%）** です。

保存した成果物を読み直して `display()` で表示します。画面の表とダウンロードしたファイルの内容を照らし合わせてください。


```python
try:
    from IPython.display import display
except ImportError:
    display = print  # Colab以外の環境向け
# 回答CSVを読み込む
answers=pd.read_csv(src,encoding='utf-8-sig')
# 満足度ごとの件数を数える
counts=answers['満足度'].value_counts().sort_index().rename_axis('満足度').reset_index(name='件数')
# 各件数を総回答数で割る
counts['割合']=counts['件数']/len(answers)
counts['平均点']=answers['満足度'].mean()
out=base/'Tips08_アンケート集計.xlsx'; counts.to_excel(out,index=False)
# 件数の棒グラフを作る
ax=counts.plot.bar(x='満足度',y='件数',legend=False,title='Survey responses')
ax.figure.tight_layout(); chart=base/'Tips08_回答グラフ.png'; ax.figure.savefig(chart)
print(counts)

# Excelを読み直して、件数と割合の全行を確認する
display(pd.read_excel(out))
# 同時に作成したグラフ画像も表示する
from PIL import Image
with Image.open(chart) as chart_image:
    display(chart_image.copy())
```

## 3. 結果の確認

`assert` は条件が正しいか確認する命令です。条件が違えばそこで止まるため、想定外の集計を見つけられます。

件数の合計は **10件**、割合の合計は **1.0（100%）** です。


```python
# 件数や合計が期待値と一致するか検査する
assert counts['件数'].sum()==len(answers)==10
assert abs(counts['割合'].sum()-1)<1e-9
print('確認OK:',out,chart)
```

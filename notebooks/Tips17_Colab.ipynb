# Tips 17：Webページの表をExcelに取り込む

**業務の困りごと**：公開Web表を手作業でExcelにコピーする。

**このTipsでできること**：`pd.read_html` はHTMLの `<table>` を表として取り出します。最初の表を選び、Excelへ書き出します。

**使用ライブラリ**：pandas

### ファイルの場所
`/content/` はColabの一時領域です。PCやGoogle Driveには自動保存されません。左側のフォルダから結果を選んでダウンロードできます。セッション終了前に保存してください。

上から順番に実行してください。各Tipsのサンプルはノートブック内で作ります。


## 1. 練習用データと準備

外部アクセスなしで、商品と売上のHTML表を文字列として用意します。

最後の `display()` で練習用データ全体を表として表示します。列名・行数・値を確認してから次へ進みましょう。


```python
try:
    from IPython.display import display
except ImportError:
    display = print  # Colab以外の環境向け
import pandas as pd
from io import StringIO
from pathlib import Path
base=Path('/content') if Path('/content').exists() else Path.cwd()
# 外部通信を使わない練習用HTML表
html='<table><tr><th>商品</th><th>売上</th></tr><tr><td>茶</td><td>1200</td></tr><tr><td>菓子</td><td>900</td></tr></table>'
print('HTMLに含まれる表（練習用データ）:')
display(pd.read_html(StringIO(html))[0])
```

## 2. 自動処理

`pd.read_html` はHTMLの `<table>` を表として取り出します。最初の表を選び、Excelへ書き出します。

**結果を見るポイント**：2商品の売上合計が **2,100円** です。

保存した成果物を読み直して `display()` で表示します。画面の表とダウンロードしたファイルの内容を照らし合わせてください。


```python
try:
    from IPython.display import display
except ImportError:
    display = print  # Colab以外の環境向け
# HTMLの最初のtable要素を表として読み込む
frame=pd.read_html(StringIO(html))[0]
# Excelへ出力する
out=base/'Tips17_Web表.xlsx'; frame.to_excel(out,index=False)
print(frame)

# 保存したExcelを読み直して全行を表示する
display(pd.read_excel(out))
```

## 3. 結果の確認

`assert` は条件が正しいか確認する命令です。条件が違えばそこで止まるため、想定外の集計を見つけられます。

2商品の売上合計が **2,100円** です。


```python
# 件数や合計が期待値と一致するか検査する
assert frame['売上'].sum()==2100
print('確認OK:',out)
```

### ファイルの場所

上から順番に実行してください。サンプルは各ノートブック内で作ります。再実行時は同名の練習ファイルを上書きします。


## 1. 練習用データと準備
サンプルを生成し、`head()` で先頭の内容を確認します。実務では、この入力を自分のファイルに置き換えます。


```python
import pandas as pd
from pathlib import Path
base=Path('/content') if Path('/content').exists() else Path.cwd()
src=base/'tips19_売上.csv'
pd.DataFrame({'店舗':['鹿児島','霧島'],'売上':[1200,900]}).to_csv(src,index=False,encoding='utf-8-sig')
print('まずは一時領域:',src)
print('練習用CSV（先頭5行）:')
print(pd.read_csv(src,encoding='utf-8-sig').head())
```

## 2. 自動処理
入力を読み込み、目的の形式で保存します。


```python
frame=pd.read_csv(src,encoding='utf-8-sig')
out=base/'Tips19_売上集計.csv'
frame.groupby('店舗',as_index=False)['売上'].sum().to_csv(out,index=False,encoding='utf-8-sig')
print(pd.read_csv(out))
```

## 3. 件数・内容の確認
入力と出力が意図した内容か確認します。


```python
assert out.exists() and pd.read_csv(out)['売上'].sum()==2100
print('確認OK')
```

## 発展：Google Driveへ保存（Colab専用・任意）
認証画面で本人のGoogleアカウントを選びます。`MyDrive/ColabTips` を作成してコピーします。


```python
# ColabでDriveへ保存したい場合だけ実行
from google.colab import drive
drive.mount('/content/drive')
import shutil
destination=Path('/content/drive/MyDrive/ColabTips')
destination.mkdir(exist_ok=True)
shutil.copy2(out,destination/out.name)
print('Driveへ保存:',destination/out.name)
```

## 4. 持ち帰り・練習

**練習問題**：Drive接続セルを実行し、MyDriveの自分専用フォルダへ結果をコピーしてみましょう。

次のセルはColab上で本人の認証後にだけ実行してください。

出力をPCへ保存したい場合は次のセルをColabで実行します。Colab以外では出力場所を表示します。

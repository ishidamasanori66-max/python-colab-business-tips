## 1. 練習用データと準備
サンプルを生成します。実務では、この入力を自分のファイルに置き換えます。


```python
import pandas as pd
from pathlib import Path
base=Path('/content') if Path('/content').exists() else Path.cwd()
src=base/'tips04_顧客.csv'
pd.DataFrame({'顧客ID':[1,2,2,3,None],'氏名':['田中','佐藤','佐藤','山田',None],'状態':['有効','有効','有効','退会',None]}).to_csv(src,index=False,encoding='utf-8-sig')
```

## 2. 自動処理
入力を読み込み、目的の形式で保存します。


```python
df=pd.read_csv(src,encoding='utf-8-sig')
clean=(df.dropna(subset=['顧客ID','氏名']).drop_duplicates(subset=['顧客ID']).query('状態 == "有効"').copy())
clean['顧客ID']=clean['顧客ID'].astype(int)
out=base/'Tips04_有効顧客.csv'; clean.to_csv(out,index=False,encoding='utf-8-sig')
print('処理前',len(df),'処理後',len(clean)); print(clean)
```

## 3. 件数・内容の確認
入力と出力が意図した内容か確認します。


```python
assert len(clean)==2 and clean['顧客ID'].is_unique
print('確認OK:',out)
```

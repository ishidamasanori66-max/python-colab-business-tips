# Tips 12：PDFから文字を抽出

**業務の困りごと**：PDFを1枚ずつ開いて文字をコピーする。

**このTipsでできること**：`PdfReader` でページを読み、`extract_text` で文字を取り出してUTF-8のテキストに保存します。

**使用ライブラリ**：pypdf, reportlab

### ファイルの場所
`/content/` はColabの一時領域です。PCやGoogle Driveには自動保存されません。左側のフォルダから結果を選んでダウンロードできます。セッション終了前に保存してください。

上から順番に実行してください。各Tipsのサンプルはノートブック内で作ります。


## 1. 練習用データと準備

文字情報を持つ1ページのPDFを作ります。画像だけのPDFとは異なります。

最後の `head()` で先頭の行を表として表示します。列名・行数・値を確認してから次へ進みましょう。


```python
%pip -q install pypdf reportlab
from pathlib import Path
from reportlab.pdfgen import canvas
from pypdf import PdfReader
base=Path('/content') if Path('/content').exists() else Path.cwd()
# 抽出元のPDFの保存先
src=base/'tips12_report.pdf'
# 文字情報を含む練習用PDFを作る
c=canvas.Canvas(str(src)); c.drawString(80,740,'Monthly report: sales 1200 JPY'); c.save()
import pandas as pd
print('抽出対象のPDF（先頭5ページ）:')
pd.DataFrame({'ページ番号':range(1,len(PdfReader(src).pages)+1),'文字の例':[(p.extract_text() or '')[:80] for p in PdfReader(src).pages]}).head()
```

## 2. 自動処理

`PdfReader` でページを読み、`extract_text` で文字を取り出してUTF-8のテキストに保存します。

**結果を見るポイント**：結果に `Monthly report` が含まれることを確認します。


```python
# PDFを読み込む
reader=PdfReader(src)
# 各ページから抽出した文字を改行でつなぐ
text='\n'.join(page.extract_text() or '' for page in reader.pages)
# 抽出した文字をテキストファイルへ保存する
out=base/'Tips12_抽出結果.txt'; out.write_text(text,encoding='utf-8')
print(text)
```

## 3. 結果の確認

`assert` は条件が正しいか確認する命令です。条件が違えばそこで止まるため、想定外の集計を見つけられます。

結果に `Monthly report` が含まれることを確認します。


```python
# 件数や合計が期待値と一致するか検査する
assert 'Monthly report' in text
print('確認OK:',out)
```

# Tips 14：PowerPointを自動作成

**業務の困りごと**：定例会の同じ構成のスライドを毎月作る。

**このTipsでできること**：`Presentation` で新しい資料を作り、タイトルスライドと商品別売上スライドを順に追加します。

**使用ライブラリ**：python-pptx

### ファイルの場所
`/content/` はColabの一時領域です。PCやGoogle Driveには自動保存されません。左側のフォルダから結果を選んでダウンロードできます。セッション終了前に保存してください。

上から順番に実行してください。各Tipsのサンプルはノートブック内で作ります。


## 1. 練習用データと準備

商品別売上2件を、定例会用のスライドに載せます。

最後の `head()` で先頭の行を表として表示します。列名・行数・値を確認してから次へ進みましょう。


```python
%pip -q install python-pptx
from pptx import Presentation
from pptx.util import Inches,Pt
from pathlib import Path
base=Path('/content') if Path('/content').exists() else Path.cwd()
# スライドに載せる商品と売上
data={'茶':1200,'菓子':900}
import pandas as pd
print('スライドに使う元データ（先頭5行）:')
pd.DataFrame(data.items(),columns=['商品','売上金額']).head()
```

## 2. 自動処理

`Presentation` で新しい資料を作り、タイトルスライドと商品別売上スライドを順に追加します。

**結果を見るポイント**：完成ファイルは **2枚** のスライドです。PowerPointで開いて文字の収まりも確認しましょう。


```python
# 新しいPowerPointを作る
prs=Presentation()
# 1枚目のタイトルスライドを作る
slide = prs.slides.add_slide(prs.slide_layouts[0])
slide.shapes.title.text = '月次売上報告'
slide.placeholders[1].text='2026年9月'
slide = prs.slides.add_slide(prs.slide_layouts[1])
slide.shapes.title.text = '商品別売上'
# 2枚目に商品ごとの売上を並べる
slide.placeholders[1].text='\n'.join(f'{k}: {v:,} 円' for k,v in data.items())
# PowerPointファイルを保存する
out=base/'Tips14_月次報告.pptx'; prs.save(out)
print(out)
```

## 3. 結果の確認

`assert` は条件が正しいか確認する命令です。条件が違えばそこで止まるため、想定外の集計を見つけられます。

完成ファイルは **2枚** のスライドです。PowerPointで開いて文字の収まりも確認しましょう。


```python
# 件数や合計が期待値と一致するか検査する
assert len(Presentation(out).slides)==2
print('確認OK')
```

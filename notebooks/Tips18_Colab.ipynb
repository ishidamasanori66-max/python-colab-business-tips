# Tips 18：Web情報を取得してCSV化

**業務の困りごと**：Web上の公開記事情報を一覧へ転記する。

**このTipsでできること**：BeautifulSoupの `select("article")` で記事ごとに取り出し、見出しと本文を抽出してCSVにします。

**使用ライブラリ**：requests, BeautifulSoup

## 1. 練習用データと準備

2件の記事を含む練習用HTMLを文字列として用意します。外部サイトへはアクセスしません。

最後の `display()` で練習用データ全体を表として表示します。列名・行数・値を確認してから次へ進みましょう。


```python
try:
    from IPython.display import display
except ImportError:
    display = print  # Colab以外の環境向け
%pip -q install beautifulsoup4 requests
from bs4 import BeautifulSoup
import pandas as pd
from pathlib import Path
base=Path('/content') if Path('/content').exists() else Path.cwd()
# 記事が2件含まれる練習用HTML
html='<article><h2>お知らせ</h2><p>新商品を発売</p></article><article><h2>営業日</h2><p>日曜は休み</p></article>'
print('HTMLに含まれる記事（練習用データ）:')
display(pd.DataFrame([{'見出し':a.h2.get_text(strip=True),'内容':a.p.get_text(strip=True)} for a in BeautifulSoup(html,'html.parser').select('article')]))
```

## 2. 自動処理

BeautifulSoupの `select("article")` で記事ごとに取り出し、見出しと本文を抽出してCSVにします。

**結果を見るポイント**：CSVは **2行** です。見出しと本文が対応しているか確認しましょう。

保存した成果物を読み直して `display()` で表示します。画面の表とダウンロードしたファイルの内容を照らし合わせてください。


```python
try:
    from IPython.display import display
except ImportError:
    display = print  # Colab以外の環境向け
# HTMLを解析できる形にする
soup=BeautifulSoup(html,'html.parser')
# 各記事の見出しと本文を取り出す
rows = []
for article in soup.select('article'):
    rows.append({
        '見出し': article.h2.get_text(strip=True),
        '内容': article.p.get_text(strip=True),
    })
# 一覧をCSVとして保存する
out=base/'Tips18_記事一覧.csv'; pd.DataFrame(rows).to_csv(out,index=False,encoding='utf-8-sig')
print(rows)

# 保存後のCSVを読み直して全行を表示する
display(pd.read_csv(out, encoding='utf-8-sig'))
```

## 3. 結果の確認

`assert` は条件が正しいか確認する命令です。条件が違えばそこで止まるため、想定外の集計を見つけられます。

CSVは **2行** です。見出しと本文が対応しているか確認しましょう。


```python
# 件数や合計が期待値と一致するか検査する
assert len(rows)==2 and out.is_file()
print('確認OK')
```
